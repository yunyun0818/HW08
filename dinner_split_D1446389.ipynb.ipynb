```python
# 原始資料輸入
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
service_rate = 0.10
discount = 100
budget_per_person = 300

# 1. 整理聚餐名稱：去除頭尾空白並轉為小寫
cleaned_event_name = event_name.strip().lower()

# 2. 整理餐費與人數並轉換型別
# 移除 "NT$"、"," 與空白，並轉為 float 數字
cleaned_meal_str = meal_amount.replace("NT$", "").replace(",", "").strip()
float_meal_amount = float(cleaned_meal_str)

# 將文字人數轉為整數
int_people = int(people)

# 3. 分步計算費用
service_fee = float_meal_amount * service_rate  # 服務費：以原始餐費計算
total_amount = float_meal_amount + service_fee - discount  # 應付總額
per_person_fee = total_amount / int_people  # 每人分攤費用

# 4. 判斷是否超過預算（布林變數）
is_over_budget = per_person_fee > budget_per_person

# 5. 輸出費用摘要
print("--- 測試一執行結果 ---")
print(f"聚餐名稱：{cleaned_event_name}")
print(f"原始餐費：{float_meal_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{int_people} 人")
print(f"每人費用：{per_person_fee:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```
